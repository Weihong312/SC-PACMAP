## Evaluation measuresK-means receives each raw 2-D embedding directly, with no additional standardization, normalization, rotation, or alignment. Spatial continuity uses each interval's 10 nearest non-self raw-XYZ neighbours:$$\frac{H}{N} = -\frac{1}{N}\sum_{i=1}^{N}\sum_j p_{ij}\log p_{ij}.$$Geochemical delineation is measured in the standardized feature space:$$\frac{\mathrm{WCSS}}{\mathrm{TSS}} = \frac{\sum_k\sum_{i \in k}\lVert x_i-\bar{x}_k\rVert^2}{\sum_i\lVert x_i-\bar{x}\rVert^2}.$$Lower entropy and WCSS/TSS are preferable; ARI is a held-out lithology reference only.
